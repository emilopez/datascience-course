```python
import polars as pl
import pandas as pd
import plotly.graph_objects as go

from pathlib import Path
```

```python
path_pard_l1 = Path.cwd().parent / "datos" / "paradasl1.csv"
paradasl1 = pl.read_csv(path_pard_l1)
```

```python
paradasl1.head()
```

```python
x = paradasl1["FacilityID"]
```

```python
y_total20 = paradasl1["V20L1t0"]
y_total18 = paradasl1["V18L1t0"]
y_total22 = paradasl1["V22L1t0"]
```

```python
fig = go.Figure()

fig.add_trace(
    go.Bar(x=x, y=y_total20, name="Dia 20/03/2023")
)
fig.add_trace(
    go.Bar(x=x, y=y_total18, name="Dia 18/03/2024")
)

fig.show()
```

```python
fig = go.Figure()
ESCALA = 20
fig.add_trace(
    go.Scatter(x=x, 
               y=[1]*len(y_total20), 
               mode="markers", 
               name="Dia 20/03/2023",
               marker=dict(size=y_total20/ESCALA,),
               customdata=y_total20,
               hovertemplate='<br>Validaciones: %{customdata}<extra></extra>'
    )
)
fig.add_trace(
    go.Scatter(x=x, 
               y=[2]*len(y_total18), 
               mode="markers", 
               name="Dia 18/03/2024",
               marker=dict(size=y_total18/ESCALA,),
               customdata=y_total18,
               hovertemplate='<br>Validaciones: %{customdata}<extra></extra>'
    )
)

fig.add_trace(
    go.Scatter(x=x, 
               y=[3]*len(y_total22), 
               mode="markers", 
               name="Dia 22/05/2024",
               marker=dict(size=y_total22/ESCALA,),
               customdata=y_total22,
               hovertemplate='<br>Validaciones: %{customdata}<extra></extra>'
    )
)
fig.update_xaxes(dtick=10)

fig.show()
```
