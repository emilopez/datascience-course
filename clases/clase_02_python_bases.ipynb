## Archivos de texto en python

Vamos a cargar en nuestro programa el contenido de un archivo, esta acción es denominada lectura. Para eso son necesarias dos acciones:
- apertura
- lectura
- cierre

Para abrirlo usamos la función ``open`` y entre comillas debemos poner la ruta al archivo al que pretendemos acceder.
**OJO: dependiendo de tu sistema operativo es el formato de la ruta que debemos poner (YA NO MAS SI USAMOS PATHLIIB)**

````python
# IMPORTAMOS de la lib estándar
from pathlib import Path

# esto depende de la ruta de cada archivo
ruta_actual = Path.cwd() # devuelve path del presente notebook
path_archi_cumples = ruta_actual.parent / "datos" / "queloscumplas.txt"

# 1) abrimos el archivo
archi_cumples = open(path_archi_cumples, 'r', encoding='utf-8')

# 2) leemos TODO su contenido y lo almacenamos en un contenedor
contenido_cumples = archi_cumples.readlines()

# 3) cerramos el archivo, ya tenemos su contenido
archi_cumples.close()
````

En contenido tenemos un contenedor, accedemos a cada elemento poniendo entre corchetes una posición (índice numérico). El primero corresponde a la posición 0, por lo tanto si son N renglones, tendremos entre 0 y N-1.

````python
type(contenido_cumples)
````

````python
# accedemos al elemento
contenido_cumples[0]
````

````python
# con índice negativo accedemos al último
contenido_cumples[-1]
````

````python
# vemos la cantidad de elementos de la lista
# en este caso equivale a la cantidad de renglones del archivo
n = len(contenido_cumples)

# con esto obtengo el elemento de la mitad, 
# la // es el operador de divisón entera
mitad = n // 2 - 1
contenido_cumples[mitad]
````

````python
# otra forma de acceder al último elemento
contenido_cumples[n-1]
````

## Ciclos repetitivos/iterativos

Como la vida es corta necesitamos ciclos repetitivos. Veamos uno a uno todos los elementos de la lista que creamos previamente:

````python
meses = "enero febrero"
for letra in meses:
    print(letra)
````

````python
for fecha_hora in contenido_cumples[0:10]:
    print(fecha_hora, end="")
````

````python
for elemento in lista_variada:
    print(elemento)
print("esto")
````

El ciclo previo se denomina ``for`` y va asignando uno a uno los elementos de la lista curso a la variable ``elemento``. 
La particularidad de **Python** es que cada acción que deba ejecutarse dentro del ciclo for se indica con **cuatro espacios**. En el siguiente ejemplo, vemos que una vez que termina el ciclo muestra se ejecuta la línea ``print("Arlt")``.

````python
for elemento in lista_variada:
    print(elemento)
print("Arlt")
````

**Observemos atentamente** el cambio que se produce si dejamos 4 espacios en el ``print`` y por ende pasa a estar dentro del ciclo repetitivo:

````python
for elemento in lista_variada:
    print(elemento)
    print("Arlt")
````

## Parseando datos

Vamos a excluir y/o seleccionar datos del archivo usando el método ``split()`` que se aplica a cadenas (``str``):

````python
for renglon in contenido_cumples[0:10]:
    fecha, hora = renglon.split()
    h,m,s = hora.split(":")
    print(fecha, s)
````

````python
for una_fecha in contenido_cumples[:10]:
    anio, mes, todo = una_fecha.split("-")
    print(mes)
````

````python
en_diciembre = 0
for una_fecha in contenido_cumples:
    anio, mes, todo = una_fecha.split("-")
    if mes == "12":
        en_diciembre = en_diciembre + 1
print(en_diciembre)
````

- Estrategia general

    - paso 1: crear el diccionario de provincias inicializadas en 0
    - paso 2: contar las localidades de cada provincia
    - paso 3: identificar la provincia con mas localidades

**Paso 1**
```python
for una_loca in localidades[1:]:
    provincia = una_loca.split(",")[-1].strip()
    contador_prov[provincia] = 0
```

**Paso 2**
```python
for una_loca in localidades[1:]:
    provincia = una_loca.split(",")[-1].strip()
    contador_prov[provincia] += 1 
```

````python
# paso 1
contador_prov = dict()
for una_loca in localidades[1:]:
    provincia = una_loca.split(",")[-1].strip()
    contador_prov[provincia] = 0

# paso 2
for una_loca in localidades[1:]:
    provincia = una_loca.split(",")[-1].strip()
    contador_prov[provincia] += 1
````

````python
contador_prov
````

````python
cant_mayor = 0
prov_mayor = ""
for prov,cantidad in contador_prov.items():
    if cantidad > cant_mayor:
        cant_mayor = cantidad
        prov_mayor = prov
        
print(prov_mayor,cant_mayor)
````

````python
max_key = max(contador_prov, key=contador_prov.get)
print(max_key, contador_prov[max_key])  # 'b'
````
