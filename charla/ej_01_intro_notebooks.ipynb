## Mi primer plot interactivo - Analisis exploratorio de los datos (AED)

```python
import numpy as np
import plotly.graph_objects as go
```

```python
# voy a crear un conjunto de datos para X
f = 1
X = np.arange(0,2,0.01)
Y = np.sin(2*np.pi*f*X)
```

```python
X
```

```python
Y
```

```python
fig = go.Figure()
fig.add_trace(go.Scatter(x=X, y=Y, mode="markers+lines"));
```

```python
Y2 = 2*Y
```

```python
fig.add_trace(go.Scatter(x=X, y=Y2, mode="markers+lines"));
```

```python
fig.show()
```
